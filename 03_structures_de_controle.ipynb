## 1. if, elif, else

Le cas le plus simple : une condition, un bloc à exécuter si elle est vraie.

````python
x = 1
if x > 0:
    print(x, 'positif')
````

La structure est toujours la même : le mot-clé, la condition, deux points, puis
le bloc indenté.

Si la condition est fausse, le bloc est simplement sauté et rien ne s'affiche.
Essaie avec `x = -1`.

### else : le cas contraire

````python
x = -1
if x > 0:
    print(x, 'positif')
else:
    print(x, 'négatif')
````

`else` n'a pas de condition : il attrape tout ce que le `if` n'a pas pris.

Mais ce code contient une imprécision. Avec `x = 0`, il affiche « négatif »,
ce qui est faux. Il manque un cas.

### elif : les cas intermédiaires

````python
x = 0

if x > 0:
    print(x, 'positif')
elif x == 0:
    print(x, 'nul')
else:
    print(x, 'négatif')
````

`elif` est la contraction de *else if*. On peut en enchaîner autant qu'on veut.

Le point important : Python s'arrête au premier bloc vrai. Les suivants ne
sont même pas testés.

````python
x = 5
if x > 0:
    print('premier bloc')
elif x > 3: # vrai aussi, mais jamais atteint
    print('deuxième bloc')
else:
    print('troisième bloc')
````

`x > 3` est vrai, pourtant le deuxième bloc ne s'exécute pas. Le premier a déjà
été retenu.

Conséquence pratique : l'ordre des conditions compte. Il faut placer les cas
les plus spécifiques en premier, sinon un cas général les avale.

````python
note = 18

# Mauvais ordre : tout passe par le premier cas
if note >= 10:
    mention = 'admis'
elif note >= 16: # jamais atteint
    mention = 'très bien'

print('Mauvais ordre :', mention)

# Bon ordre : du plus exigeant au plus général
if note >= 16:
    mention = 'très bien'
elif note >= 10:
    mention = 'admis'
else:
    mention = 'refusé'
print('Bon ordre    :', mention)
````

## 2. Les conditions et l'indentation

Une condition est simplement une expression qui vaut `True` ou `False`. On peut
l'évaluer seule, sans `if`.

````python
x = 0
print(x > 0)
print(x == 0)
print(type(x > 0))
````

Le `if` ne fait rien de plus que regarder cette valeur. On peut donc lui donner
un booléen directement :

````python
if True:
    print('toujours exécuté')
if False:
    print('jamais exécuté')
````

`if True:` a l'air inutile, mais c'est un bon outil de débogage : ça permet de
forcer temporairement un chemin pour tester un bloc.

### Combiner des conditions

C'est ici qu'on retrouve le piège du notebook précédent.

````python
x = 1
y = 2
if (x > 0) and (y > x):
    print(x, y, 'positifs et croissants')
````

En Python pur, l'opérateur logique est **`and`**, pas `&`.

`&` fonctionne aussi et donne le même résultat sur des booléens, mais ce n'est
pas le même opérateur : `&` travaille bit à bit, et il évalue toujours les deux
côtés.

La différence devient visible ici :

````python
liste = []

# Avec and : sûr
if liste and liste[0] > 5:
    print('le premier élément dépasse 5')
else:
    print('and : pas de plantage sur une liste vide')

# Avec & : IndexError, car liste[0] est évalué quand même
try:
    if liste & (liste[0] > 5):
        pass
except IndexError as e:
    print('& : IndexError -', e)
````

`and` s'arrête dès qu'il sait conclure : la liste étant vide donc fausse, il ne
regarde pas `liste[0]`. C'est l'évaluation court-circuit.

La règle : `and`, `or`, `not` en Python pur. `&`, `|`, `~` uniquement avec
NumPy et Pandas, où l'on compare des tableaux entiers.

### L'indentation

Python n'utilise ni accolades ni `end`. C'est l'indentation qui délimite les
blocs, et elle fait partie de la syntaxe.

````python
x = -5

if x > 0:
    print('dans le if')
    print('encore dans le if')
print('hors du if') # non indenté : toujours exécuté
````

La troisième ligne s'affiche même si la condition est fausse, parce qu'elle est
sortie du bloc.

Une indentation incohérente lève une erreur :

```python
if x > 0:
    print('a')
      print('b')      # IndentationError
```

La convention officielle est **quatre espaces**. Les tabulations fonctionnent
mais mélanger les deux provoque des erreurs difficiles à voir, puisqu'elles sont
invisibles à l'écran.

### Dans une fonction

Les conditions s'utilisent naturellement dans une fonction. Attention à
l'indentation, qui a maintenant deux niveaux.

````python
def signe(x):
    """Affiche le signe d'un nombre."""
    if x > 0:
        print(x, 'positif')
    elif x == 0:
        print(x, 'nul')
    else:
        print(x, 'négatif')

signe(-5)
signe(0)
signe(1)
````

### Le piège de print et return

Cette fonction affiche mais ne renvoie rien. C'est le point vu au notebook
précédent, et il produit ici un résultat déroutant.

````python
print(signe(5))
````

Deux lignes s'affichent : le `print` interne de la fonction, puis `None`, parce
que `signe` ne renvoie rien et que le `print` externe affiche cette absence.

Dans une boucle, l'effet est spectaculaire on obtient un `None` après chaque
ligne utile. C'est un signal clair qu'il faut un `return`.

````python
def signe_ok(x):
    """Retourne le signe d'un nombre, sous forme de texte."""
    if x > 0:
        return 'positif'
    elif x == 0:
        return 'nul'
    else:
        return 'négatif'

print(signe_ok(5))
print(f'{-3} est {signe_ok(-3)}')

resultats = [signe_ok(n) for n in [-2, 0, 7]]
print(resultats)
````

La version avec `return` est réutilisable : dans une f-string, dans une liste,
dans une autre fonction. La version avec `print` ne sert qu'à afficher.

À noter : `return` sort immédiatement de la fonction. Les blocs suivants ne
sont pas évalués, ce qui rend le `elif` presque superflu ici mais le garder
reste plus lisible.

## 3. La boucle for

`for` répète un bloc pour chaque élément d'une séquence.

````python
for i in range(5):
    print('Bonjour')
````

La variable `i` existe et change à chaque tour, même si on ne s'en sert pas.
Affichons-la :

````python
for i in range(5):
    print(i)
````

Le nom de la variable est libre. `i` est une convention pour un compteur, mais
un nom parlant vaut mieux dès que le contenu a un sens.

````python
for element in range(3):
    print(element)

for prenom in ['Ada', 'Alan', 'Grace']:
    print('Bonjour', prenom)
````

Par convention, on utilise `_` quand la variable ne sert pas :

```python
for _ in range(3):
    print('trois fois')
```

C'est un nom de variable comme un autre, mais il signale au lecteur qu'on ignore
délibérément la valeur.

````python
for _ in range(3):
    print('trois fois')
````

## 4. range et ses trois arguments

`range` génère une suite de nombres. Elle accepte un, deux ou trois arguments.

````python
print('range(5)        ->', list(range(5)))
print('range(2, 7)     ->', list(range(2, 7)))
print('range(0, 10, 2) ->', list(range(0, 10, 2)))
print('range(10, 0, -1)->', list(range(10, 0, -1)))
````

| Forme | Signification |
|---|---|
| `range(fin)` | de 0 à fin-1 |
| `range(début, fin)` | de début à fin-1 |
| `range(début, fin, pas)` | avec un pas, éventuellement négatif |

La borne de fin est toujours exclue. `range(5)` s'arrête à 4, `range(2, 7)`
s'arrête à 6.

C'est cohérent avec le reste de Python : le slicing `liste[2:7]` exclut aussi
l'indice 7. La logique est que `range(n)` produit exactement `n` valeurs.

### Le pas négatif

````python
for element in range(10, -10, -5):
    print(element, end=' ')
````

Avec un pas négatif, le début doit être plus grand que la fin. Sinon la boucle
ne tourne pas du tout, sans erreur :

````python
print('range(0, 10, -1) ->', list(range(0, 10, -1))) # vide
print('range(10, 0)     ->', list(range(10, 0))) # vide aussi
````

Deux listes vides, aucun message. Si une boucle ne s'exécute jamais alors qu'elle
le devrait, c'est la première chose à vérifier.

### range ne fabrique pas la liste

`range` ne stocke pas les nombres : il les calcule au fur et à mesure. C'est
pour ça qu'on doit écrire `list(range(5))` pour les voir.

````python
import sys

print(range(10))
print('range(10)        :', sys.getsizeof(range(10)), 'octets')
print('range(10 000 000):', sys.getsizeof(range(10_000_000)), 'octets')
print('list(range(10))  :', sys.getsizeof(list(range(10))), 'octets')
````

Un `range` sur dix millions de valeurs occupe la même place qu'un `range` sur
dix : il ne retient que le début, la fin et le pas.

C'est ce qui permet d'écrire `for i in range(10_000_000)` sans saturer la
mémoire. Construire la liste correspondante prendrait plusieurs centaines de
mégaoctets.

L'underscore dans `10_000_000` est un séparateur de milliers, ignoré par Python.
Purement pour la lisibilité.

````python
for i in range(10_000_000):
  print(i)
````

## 5. Parcourir autre chose que des nombres

`for` ne se limite pas à `range`. Il parcourt tout ce qui est **itérable** :
listes, chaînes, dictionnaires, fichiers, tableaux NumPy.

````python
for lettre in 'Python':
    print(lettre, end=' ')
print()

for note in [12, 15, 8, 19]:
    print(note, 'sur 20')
````

### L'erreur du débutant

Quand on vient d'un autre langage, on écrit souvent ceci :

````python
notes = [12, 15, 8, 19]

# Version détournée
for i in range(len(notes)):
    print(notes[i])
````

Ça marche, mais c'est un détour. On fabrique des indices pour aller chercher des
éléments, alors que `for` sait parcourir directement :

```python
for note in notes:
    print(note)
```

Plus court, plus lisible, et sans risque d'erreur d'indice.

### Quand on a besoin de l'indice

Si l'indice est vraiment nécessaire, `enumerate` le fournit sans avoir à gérer
un compteur.

````python
for i, note in enumerate(notes):
    print(f'Note numéro {i} : {note}')
````

`enumerate` renvoie des couples `(position, valeur)` que la boucle déballe. C'est
exactement ce qui servait à colorer les styles de ligne dans le notebook
Matplotlib.

### Parcourir deux listes en parallèle

`zip` assemble deux séquences en couples.

````python
prenoms = ['Ada', 'Alan', 'Grace']
metiers = ['mathématicienne', 'logicien', 'informaticienne']

for prenom, metier in zip(prenoms, metiers):
    print(f'{prenom} était {metier}')
````

`zip` s'arrête à la plus courte des deux séquences, sans avertir. Si tes listes
n'ont pas la même longueur, des éléments disparaissent silencieusement.

### Les boucles imbriquées

Une boucle peut en contenir une autre. La boucle intérieure fait un tour complet
à chaque itération de l'extérieure.

````python
for i in range(1, 4):
    for j in range(1, 4):
        print(f'{i} x {j} = {i*j}', end='   ')
    print() # saut de ligne à la fin de chaque ligne du tableau
````

Neuf itérations au total : trois fois trois. C'est le mécanisme qui servait à
produire la grille de nuages de points dans le notebook Matplotlib.

Attention au coût : deux boucles de mille tours font un million d'itérations.

## 6. La boucle while

`while` répète tant qu'une condition reste vraie. Contrairement à `for`, elle ne
parcourt rien : elle teste.

````python
x = 0

while x < 5:
    print(x)
    x += 1
````

Trois éléments indispensables, et l'oubli du troisième est fatal :

1. une variable initialisée avant la boucle,
2. une condition qui la teste,
3. une modification de cette variable dans la boucle.

Sans la troisième, la condition reste vraie pour toujours :

```python
x = 0
while x < 5:
    print(x) # x ne change jamais -> boucle infinie
```

Ce code tourne sans fin. Dans Colab, il faut interrompre l'exécution avec le
bouton stop, ou *Exécution* → *Interrompre l'exécution*.

### Une sécurité pendant l'apprentissage

Quand on écrit un `while` dont on n'est pas sûr, un compteur de sécurité évite de
bloquer le noyau.

````python
x = 0
tours = 0

while x < 5 and tours < 1000: # garde-fou
    x += 1
    tours += 1

print(f'Terminé en {tours} tours, x = {x}')
````

### for ou while ?

| Situation | Boucle |
|---|---|
| Le nombre de tours est connu à l'avance | `for` |
| On parcourt une liste, un fichier, une chaîne | `for` |
| On répète jusqu'à ce qu'une condition change | `while` |
| On attend une saisie ou un événement | `while` |

En pratique, `for` couvre la grande majorité des cas. `while` sert quand on ne
sait pas combien de tours seront nécessaires : une recherche, une convergence,
une saisie utilisateur.

C'est exactement ce que fait `optimize.minimize` en interne il itère jusqu'à
ce que la variation devienne négligeable, sans savoir d'avance combien de tours
cela prendra.

## 7. break, continue et le else de boucle

Ces trois mots-clés modifient le déroulement d'une boucle.

### break : sortir immédiatement

````python
for n in range(2, 30):
    if n % 7 == 0:
        print('Premier multiple de 7 trouvé :', n)
        break # on sort, inutile de continuer
    print('  test de', n)
````

`break` interrompt la boucle entière. Utile dès qu'on cherche quelque chose et
qu'on l'a trouvé : continuer serait du temps perdu.

Dans des boucles imbriquées, `break` ne sort que de la boucle où il se trouve,
pas des deux.

### continue : passer au tour suivant

````python
for n in range(10):
    if n % 2 == 0:
        continue # on saute les pairs
    print(n, end=' ')
````

`continue` abandonne le tour en cours et passe au suivant, sans sortir de la
boucle.

Il permet d'éviter des blocs trop indentés. Ces deux versions sont équivalentes :

```python
for n in liste:
    if n is not None:
        if n > 0:
            traiter(n)

for n in liste:
    if n is None:
        continue
    if n <= 0:
        continue
    traiter(n)
```

La seconde se lit mieux quand les conditions se multiplient.

### Le else de boucle

Peu connu, et pourtant utile. Le bloc `else` d'une boucle s'exécute si la
boucle s'est terminée sans break.

````python
nombres = [3, 7, 11, 15]

for n in nombres:
    if n % 2 == 0:
        print('Nombre pair trouvé :', n)
        break
else:
    print('Aucun nombre pair dans la liste')
````

Sans ce `else`, il faudrait une variable témoin :

```python
trouve = False
for n in nombres:
    if n % 2 == 0:
        trouve = True
        break
if not trouve:
    print('Aucun nombre pair')
```

Le `else` de boucle fait la même chose en deux lignes de moins.

Le nom est mal choisi `nobreak` aurait été plus clair ce qui explique qu'on
le rencontre rarement.

### Un piège : modifier une liste pendant qu'on la parcourt

````python
nombres = [1, 2, 3, 4, 5]

for n in nombres:
    if n % 2 == 0:
        nombres.remove(n) # A ne pas faire

print('Résultat :', nombres)
````

Le 4 n'a pas été supprimé.

L'explication : la boucle avance par indice. En retirant le 2, tous les éléments
suivants reculent d'un cran, et l'élément qui prend la place du 2 est sauté.

La solution consiste à parcourir une copie, ou mieux, à construire une nouvelle
liste :

````python
nombres = [1, 2, 3, 4, 5]

impairs = [n for n in nombres if n % 2 != 0]
print('Avec une compréhension :', impairs)

nombres = [1, 2, 3, 4, 5]
for n in nombres.copy(): # ou list(nombres)
    if n % 2 == 0:
        nombres.remove(n)
print('En parcourant une copie :', nombres)
````

La première version est préférable. C'est une compréhension de liste, le
sujet du notebook 06, et elle remplace avantageusement beaucoup de boucles.

````python
# Exercice 1

def fizzbuzz(n):
    """Affiche les nombres de 1 à n, avec Fizz, Buzz ou FizzBuzz."""
    for i in range(1, n + 1):
        if i % 15 == 0: # le cas le plus spécifique en premier
            print('FizzBuzz')
        elif i % 3 == 0:
            print('Fizz')
        elif i % 5 == 0:
            print('Buzz')
        else:
            print(i)

fizzbuzz(15)
````

````python
def fizzbuzz_faux(n):
    for i in range(1, n + 1):
        if i % 3 == 0:
            print(i, '-> Fizz')
        elif i % 5 == 0:
            print(i, '-> Buzz')
        elif i % 15 == 0: # jamais atteint
            print(i, '-> FizzBuzz')

fizzbuzz_faux(15)
````

````python
def fizzbuzz_v2(n):
    for i in range(1, n + 1):
        sortie = ''
        if i % 3 == 0:
            sortie += 'Fizz'
        if i % 5 == 0:
            sortie += 'Buzz'
        print(sortie or i) # si sortie est vide, affiche i
fizzbuzz_v2(15)
````

````python
def est_premier(n):
    """Retourne True si n est un nombre premier."""
    if n < 2:
        return False # 0 et 1 ne sont pas premiers

    for diviseur in range(2, n):
        if n % diviseur == 0:
            return False  # un diviseur trouvé : pas premier
    return True  # aucun diviseur : premier
````

````python
for n in [0, 1, 2, 3, 4, 17, 25, 29]:
    print(n, '->', est_premier(n))

print([n for n in range(50) if est_premier(n)])
````

````python
def est_premier_v2(n):
    if n < 2:
        return False
    for diviseur in range(2, n):
        if n % diviseur == 0:
            break
    else: # exécuté si la boucle n'a pas été interrompue
        return True
    return False
````

````python
for n in [0, 1, 2, 3, 4, 17, 25, 29]:
    print(n, '->', est_premier_v2(n))

print([n for n in range(50) if est_premier(n)])
````

````python
import math

def est_premier_rapide(n):
    if n < 2:
        return False
    for diviseur in range(2, int(math.sqrt(n)) + 1):
        if n % diviseur == 0:
            return False
    return True
````

````python
for n in [0, 1, 2, 3, 4, 17, 25, 29]:
    print(n, '->', est_premier_rapide(n))

print([n for n in range(50) if est_premier(n)])
````

````python
import time

n = 999_983 # un grand nombre premier

debut = time.time()
est_premier(n)
lent = time.time() - debut

debut = time.time()
est_premier_rapide(n)
rapide = time.time() - debut

print(f'Version simple : {lent:.4f} s')
print(f'Version racine : {rapide:.6f} s')
print(f'Rapport : {lent/rapide:.0f}x plus rapide')
````

````python
# Exercice 3

def deviner(secret, minimum=1, maximum=1000, verbeux=True):
    """Trouve un nombre par dichotomie. Retourne le nombre d'essais."""
    essais = 0

    while True:
        essais += 1
        milieu = (minimum + maximum) // 2 # division entière

        if verbeux:
            print(f'Essai {essais} : intervalle [{minimum}, {maximum}] -> {milieu}')

        if milieu == secret:
            return essais
        elif milieu < secret:
            minimum = milieu + 1 # trop bas, on remonte
        else:
            maximum = milieu - 1 # trop haut, on descend

print('Trouvé en', deviner(742), 'essais')
````

````python
essais = [deviner(s, verbeux=False) for s in range(1, 1001)]

print('Pire cas :', max(essais))
print('Moyenne  :', round(sum(essais) / len(essais), 2))

import math
print('log2(1000) =', round(math.log2(1000), 2))
````
