## 1. Créer une liste

Des crochets, des éléments séparés par des virgules.

````python
liste_1 = [1, 4, 2, 7, 35, 84]
villes = ['Paris', 'Berlin', 'Londres', 'Bruxelles']
liste_vide = []

print(liste_1)
print(villes)
print(liste_vide, '-> longueur', len(liste_vide))
````

Une liste peut contenir n'importe quoi, y compris des types mélangés et d'autres
listes.

````python
melange = [1, 'texte', 3.14, True, None]
print(melange)

imbriquee = [liste_1, villes] # une liste de listes
print(imbriquee)
print(len(imbriquee), 'éléments')  # 2, pas 10
````

`imbriquee` ne contient que deux éléments, qui se trouvent être des listes.
Pour atteindre une valeur à l'intérieur, il faut deux paires de crochets :

````python
print(imbriquee[1]) # la deuxième liste
print(imbriquee[1][0]) # le premier élément de cette liste
````

Cette structure préfigure les tableaux à deux dimensions de NumPy. La différence
est qu'un tableau NumPy s'indexe avec `tableau[1, 0]`, ce qui est plus lisible.

### Les tuples

Un tuple s'écrit avec des parenthèses. Il ressemble à une liste, mais il est
immuable : une fois créé, on ne peut plus le modifier.

````python
tuple_1 = (1, 2, 6, 1, 7)

print(tuple_1)
print(tuple_1[0]) # la lecture fonctionne pareil
print(len(tuple_1))
````

````python
# La modification échoue
try:
    tuple_1[0] = 99
except TypeError as e:
    print('TypeError :', e)

# Alors qu'une liste accepte
liste_1[0] = 99
print('Liste modifiée :', liste_1)
````

C'est la seule différence importante, et elle a des conséquences.

Pourquoi utiliser un tuple. Quand des données ne doivent pas changer, le
tuple le garantit une erreur de code lèvera une exception au lieu de corrompre
silencieusement les données. Des coordonnées, une date, une configuration.

Un tuple peut aussi servir de clé de dictionnaire, ce qu'une liste ne peut pas
faire, précisément parce qu'il ne change pas.

Et il est légèrement plus rapide et plus léger en mémoire.

Un détail de syntaxe : c'est la virgule qui fait le tuple, pas les
parenthèses.

````python
a = (1, 2, 3)
b = 1, 2, 3 # même chose, sans parenthèses
c = (5) # ATTENTION : ce n'est pas un tuple
d = (5,) # celui-ci en est un

print(type(a), type(b))
print(type(c), '->', c)
print(type(d), '->', d)
````

`(5)` est simplement le nombre 5 entre parenthèses. Pour un tuple d'un seul
élément, la virgule finale est obligatoire.

C'est aussi ce qui explique le déballage vu au notebook 02 : `x, y = 1, 2` crée
un tuple à droite et le déballe à gauche.

### Et les chaînes de caractères

Une chaîne se comporte comme une liste de caractères, en lecture. Elle est
immuable, comme un tuple.

````python
prenom = 'Ammar'
print(prenom[0])
print(len(prenom))

try:
    prenom[0] = 'O'
except TypeError as e:
    print('TypeError :', e)
````

## 2. L'indexing

Chaque élément a une position, appelée indice. Le premier indice est 0, pas 1.

````python
villes = ['Paris', 'Berlin', 'Londres', 'Bruxelles']

print(villes[0]) # premier
print(villes[1]) # deuxième
print(villes[3]) # quatrième et dernier
````

Les indices négatifs comptent depuis la fin, en commençant à -1.

````python
print(villes[-1]) # dernier
print(villes[-2]) # avant-dernier
````

```
 Paris   Berlin  Londres  Bruxelles
   0        1        2        3
  -4       -3       -2       -1
```

`villes[-1]` est bien plus sûr que `villes[len(villes)-1]` pour prendre le
dernier élément, et ça marche même si la liste change de taille.

Un indice hors limites lève une erreur :

````python
try:
    print(villes[10])
except IndexError as e:
    print('IndexError :', e)
````

## 3. Le slicing

Le slicing extrait une portion de liste. La syntaxe est
`liste[début:fin:pas]`.

````python
print(villes[0:3])
````

La borne de fin est exclue. `[0:3]` donne les indices 0, 1 et 2, soit trois
éléments.

C'est la même convention que `range(3)`, et la logique est identique : `[0:3]`
produit exactement 3 éléments.

### Les raccourcis

Chaque partie est facultative.

````python
print(villes[:3]) # du début à l'indice 2
print(villes[2:]) # de l'indice 2 à la fin
print(villes[:]) # tout
print(villes[1:3]) # du 1 au 2
````

En pratique, on écrit `[:3]` plutôt que `[0:3]` : le zéro est sous-entendu.

### Le pas

````python
print(villes[::2]) # un élément sur deux
print(villes[1::2]) # un sur deux, en partant du deuxième
````

### Le pas négatif : inverser

C'est l'idiome le plus connu de Python.

````python
print(villes[::-1]) # la liste à l'envers
print(villes[::-2]) # à l'envers, un sur deux

prenom = 'Ammar'
print(prenom[::-1]) # marche aussi sur les chaînes
````

`[::-1]` est la façon canonique d'inverser une liste ou une chaîne. Elle crée une
copie inversée sans toucher à l'original.

Attention à ne pas la confondre avec `liste.reverse()`, qui modifie la liste sur
place et ne renvoie rien.

### Le slicing ne plante jamais

Contrairement à l'indexing, un slice hors limites ne lève pas d'erreur.

````python
print(villes[10:20])    # liste vide, pas d'IndexError
print(villes[:100])     # toute la liste
````

C'est pratique, mais ça peut masquer un bug : une liste vide inattendue passe
inaperçue. Si un traitement ne donne rien, vérifie tes bornes.

## 4. Modifier une liste

### Remplacer un élément

````python
villes = ['Paris', 'Berlin', 'Londres', 'Bruxelles']

villes[0] = 'Dublin'
print(villes)
````

Impossible avec un tuple ou une chaîne, comme vu en section 1.

### Ajouter

Trois méthodes, à ne pas confondre.

````python
villes = ['Paris', 'Berlin']

villes.append('Londres') # ajoute à la fin
print('append  :', villes)

villes.insert(1, 'Madrid') # insère à l'indice 1
print('insert  :', villes)

autres = ['Amsterdam', 'Rome']
villes.extend(autres) # ajoute chaque élément
print('extend  :', villes)
````

La différence entre `append` et `extend` est celle qu'on rate le plus souvent.

````python
a = [1, 2]
a.append([3, 4])
print('append :', a, '-> longueur', len(a))

b = [1, 2]
b.extend([3, 4])
print('extend :', b, '-> longueur', len(b))
````

`append` ajoute un élément, qui se trouve être une liste. `extend` ajoute
chaque élément de la liste.

L'opérateur `+` fait comme `extend`, mais en créant une nouvelle liste :

```python
c = [1, 2] + [3, 4]      # [1, 2, 3, 4], l'original est intact
```

### Supprimer

````python
villes = ['Paris', 'Berlin', 'Londres', 'Berlin']

retire = villes.pop(1) # par INDICE, et renvoie l'élément
print('pop(1) a retiré', retire, '->', villes)

villes.remove('Berlin') # par VALEUR, ne renvoie rien
print('remove :', villes)

del villes[0] # par indice, sans renvoyer
print('del    :', villes)
````

| | `pop(i)` | `remove(x)` | `del liste[i]` |
|---|---|---|---|
| Argument | un indice | une valeur | un indice ou une tranche |
| Renvoie | l'élément retiré | rien | rien |
| Si absent | `IndexError` | `ValueError` | `IndexError` |

`pop()` sans argument retire le dernier élément. C'est ce qui permet
d'implémenter une pile.

`remove` ne supprime que la première occurrence. S'il y a des doublons, les
autres restent.

### Le piège du None

`remove`, `append`, `sort` et les autres modifient la liste sur place et ne
renvoient rien.

````python
liste = [1, 2, 3]
resultat = liste.remove(2)

print('La liste  :', liste)
print('Le retour :', resultat) # None
````

D'où l'erreur classique :

```python
liste = liste.remove(2)      # la liste est perdue, remplacée par None
```

C'est l'inverse de Pandas, où il faut écrire `data = data.dropna()`. La
règle générale : si une méthode modifie sur place, ne pas réassigner. Si elle
renvoie une copie, réassigner.

### Vider

````python
villes = ['Paris', 'Berlin']
villes.clear()
print(villes, '-> longueur', len(villes))
````

Attention en notebook : après un `clear()`, toutes les cellules suivantes
travaillent sur une liste vide. Si tu remontes exécuter une cellule qui parcourt
`villes`, elle n'affichera plus rien sans erreur.

C'est le même problème d'état que les variables écrasées. Redéfinis la liste
après un `clear()`.

````python
villes = ['Paris', 'Berlin', 'Londres', 'Bruxelles']
print(villes)
````

## 5. Trier, compter, chercher

### Trier

````python
nombres = [2, 4, 51, 2, 4, 5, 8, 9, 9, 87, 76]
nombres.sort()
print('croissant  :', nombres)
nombres.sort(reverse=True)
print('décroissant:', nombres)
````

### sort ou sorted ?

`sort()` modifie la liste et renvoie `None`. `sorted()` laisse l'original intact
et renvoie une nouvelle liste triée.

````python
original = [3, 1, 2]
print('sort() renvoie   :', original.sort()) # None
print('la liste triée   :', original)

original = [3, 1, 2]
copie = sorted(original)
print('sorted() renvoie :', copie)
print('original intact  :', original)
````

`sorted` fonctionne sur tout ce qui est itérable tuples, chaînes,
dictionnaires et renvoie toujours une liste.

Utilise `sorted` par défaut : ne pas modifier ses données est plus sûr.

### Le piège des types mélangés

````python
try:
    [1, 'a', 2].sort()
except TypeError as e:
    print('TypeError :', e)
````

Python ne sait pas comparer un texte et un nombre. Sur une liste hétérogène, le
tri échoue.

Et le tri de chaînes est alphabétique, ce qui donne des surprises sur des
nombres écrits en texte :

````python
print(sorted(['10', '9', '100'])) # ordre alphabétique
print(sorted([10, 9, 100])) # ordre numérique
````

### Compter et chercher

````python
villes = ['Paris', 'Berlin', 'Paris', 'Londres']

print('count :', villes.count('Paris')) # nombre d'occurrences
print('index :', villes.index('Berlin')) # position de la première occurrence
print('len :', len(villes))
print('in :', 'Paris' in villes)
print('not in :', 'Rome' not in villes)
````

L'opérateur `in` est le plus utile : il teste l'appartenance sans risque
d'erreur, contrairement à `index` qui lève une `ValueError` si l'élément est
absent.

````python
ville = 'Berlin'

if ville in villes:
    print('Oui,', ville, 'est dans la liste')
else:
    print('Non')
````

### Quelques fonctions utiles

````python
nombres = [4, 8, 15, 16, 23, 42]

print('min :', min(nombres))
print('max :', max(nombres))
print('sum :', sum(nombres))
print('moyenne :', sum(nombres) / len(nombres))
````

## 6. Le piège de la copie

C'est le point le plus important du notebook, et il ne lève aucune erreur.

````python
liste_a = [1, 2, 3]
liste_b = liste_a  # on croit copier

liste_b.append(4)

print('liste_a :', liste_a)
print('liste_b :', liste_b)
````

`liste_a` a changé alors qu'on n'y a pas touché.

L'explication vient du notebook 02 : `liste_b = liste_a` ne copie rien. Les deux
noms désignent le même objet en mémoire.

````python
liste_a = [1, 2, 3]
liste_b = liste_a
liste_c = liste_a.copy() # une vraie copie

print('a et b, même objet ?', liste_a is liste_b)
print('a et c, même objet ?', liste_a is liste_c)

liste_b.append(99)
liste_c.append(-1)

print()
print('liste_a :', liste_a) # modifiée par b, pas par c
print('liste_c :', liste_c)
````

Trois façons de copier, équivalentes :

```python
copie = liste.copy()
copie = liste[:]          # slicing complet
copie = list(liste)
```

### Le cas des listes imbriquées

`copy()` ne copie qu'un niveau. Les sous-listes restent partagées.

````python
original = [[1, 2], [3, 4]]
copie = original.copy()

copie[0].append(99) # on modifie une SOUS-liste

print('original :', original)
print('copie    :', copie)
````

Les deux ont changé. Pour une copie complète, il faut `deepcopy` :

````python
import copy as copie_module

original = [[1, 2], [3, 4]]
profonde = copie_module.deepcopy(original)

profonde[0].append(99)

print('original :', original)
print('profonde :', profonde)
````

Cette distinction est la raison du `.copy()` qu'on utilise systématiquement sur
les DataFrames Pandas avant de les transformer. Sans lui, trois traitements
« différents » travailleraient tous sur les mêmes données.

### Les tuples n'ont pas ce problème

Puisqu'ils sont immuables, on ne peut pas les modifier, donc le partage est sans
danger. C'est un argument en leur faveur.

## 7. Parcourir une liste

La boucle `for` du notebook 03 s'applique directement.

````python
villes = ['Paris', 'Tokyo', 'Nice', 'Valence', 'Brest']

for ville in villes:
    print(ville)
````

### Avec l'indice : enumerate

````python
for i, ville in enumerate(villes):
    print(f'{i} : {ville}')
````

`enumerate` renvoie des couples `(indice, valeur)` que la boucle déballe. C'est
plus sûr que `for i in range(len(villes))` suivi de `villes[i]`.

### Deux listes en parallèle : zip

````python
liste_1 = [16, 224, 43, 4, 57, 4, 5, 7]

for nombre, ville in zip(liste_1, villes):
    print(nombre, ville)
````

### Le piège de zip

Regarde bien la sortie ci-dessus : cinq lignes, alors que `liste_1` en contient
huit.

````python
print('liste_1 :', len(liste_1), 'éléments')
print('villes :', len(villes), 'éléments')
print('zip :', len(list(zip(liste_1, villes))), 'couples')
print('Éléments ignorés :', liste_1[len(villes):])
````

`zip` s'arrête à la plus courte des deux séquences. Trois valeurs de `liste_1`
ont disparu, sans aucun message.

C'est un piège silencieux redoutable dès qu'on croise des données de sources
différentes : des mesures et leurs étiquettes, des prédictions et leurs valeurs
réelles. Si une liste a un élément de moins, le dernier résultat s'évapore.

La parade est de vérifier avant, ou d'utiliser le mode strict :

````python
if len(liste_1) != len(villes):
    print('Attention : longueurs différentes')

# Python 3.10+ : lève une erreur au lieu d'ignorer
try:
    for a, b in zip(liste_1, villes, strict=True):
        pass
except ValueError as e:
    print('ValueError :', e)
````

### Construire une liste dans une boucle

````python
carres = []
for n in range(1, 6):
    carres.append(n**2)

print(carres)
````

Ce schéma liste vide, boucle, `append` est tellement fréquent que Python
propose une écriture condensée, la compréhension de liste :

````python
carres = [n**2 for n in range(1, 6)]
print(carres)

pairs = [n for n in range(20) if n % 2 == 0]
print(pairs)

majuscules = [v.upper() for v in villes]
print(majuscules)
````

Même résultat, une seule ligne. C'est le sujet du notebook 06, et c'est la
syntaxe qu'on retrouve partout dans les notebooks NumPy et Pandas.

## 8. Mémo

### Liste ou tuple

| | Liste `[]` | Tuple `()` |
|---|---|---|
| Modifiable | oui | non |
| Clé de dictionnaire | non | oui |
| Usage | données qui évoluent | données fixes |

### Les méthodes de liste

| Méthode | Effet | Renvoie |
|---|---|---|
| `append(x)` | ajoute un élément à la fin | rien |
| `extend(l)` | ajoute chaque élément de `l` | rien |
| `insert(i, x)` | insère à l'indice `i` | rien |
| `pop(i)` | retire par indice | l'élément |
| `remove(x)` | retire par valeur | rien |
| `clear()` | vide la liste | rien |
| `sort()` | trie sur place | rien |
| `count(x)` | compte les occurrences | un entier |
| `index(x)` | position de la première occurrence | un entier |
| `copy()` | copie superficielle | une liste |

### Slicing

| Écriture | Effet |
|---|---|
| `l[2]` | l'élément d'indice 2 |
| `l[-1]` | le dernier |
| `l[:3]` | les trois premiers |
| `l[2:]` | à partir du troisième |
| `l[::2]` | un sur deux |
| `l[::-1]` | à l'envers |

### Les pièges

| Situation | Ce qui se passe |
|---|---|
| `b = a` sur une liste | les deux noms pointent le même objet |
| `copy()` sur des listes imbriquées | les sous-listes restent partagées |
| `a = a.remove(x)` | `a` devient `None` |
| `zip` sur des longueurs inégales | les éléments en trop sont ignorés |
| `append` au lieu de `extend` | ajoute la liste entière comme un élément |
| `sort()` sur des types mélangés | `TypeError` |
| `clear()` en notebook | les cellules suivantes voient une liste vide |

## Pour continuer

Le notebook suivant porte sur les dictionnaires, qui associent des clés à des
valeurs plutôt que des positions à des valeurs.

Puis viendra la compréhension de liste, entrevue en section 7, qui remplace
avantageusement la plupart des boucles de construction.

````python
# Exerice 1

def sans_doublons(liste):
    """Retourne une nouvelle liste sans doublons, dans l'ordre d'apparition."""
    resultat = []
    for element in liste:
        if element not in resultat: # on n'ajoute que si absent
            resultat.append(element)
    return resultat
````

````python
donnees = ['b', 'a', 'c', 'a', 'b', 'd']
propre = sans_doublons(donnees)

print('résultat :', propre)
print('original :', donnees)
print('même objet ?', propre is donnees)
````

````python
print('avec la boucle :', sans_doublons(donnees))
print('avec set       :', list(set(donnees)))
````

````python
def sans_doublons_rapide(liste):
    """Ordre préservé, et rapide."""
    return list(dict.fromkeys(liste))

print(sans_doublons_rapide(donnees)) # ['b', 'a', 'c', 'd']
````

````python
# Exercice 2
def fusionner(l1, l2):
    """Alterne les éléments de deux listes, même de longueurs différentes."""
    resultat = []
    for i in range(max(len(l1), len(l2))):
        if i < len(l1):
            resultat.append(l1[i])
        if i < len(l2):
            resultat.append(l2[i])
    return resultat
````

````python
print(fusionner([1, 2, 3], ['a', 'b'])) # [1, 'a', 2, 'b', 3]
print(fusionner(['a', 'b'], [1, 2, 3])) # ['a', 1, 'b', 2, 3]
print(fusionner([], ['a'])) # ['a']
print(fusionner([1, 2, 3], []))  # [1, 2, 3]
````

````python
resultat = []
for a, b in zip([1, 2, 3], ['a', 'b']):
    resultat.extend([a, b])
print(resultat) # [1, 'a', 2, 'b'] le 3 manque
````

````python
from itertools import zip_longest
def fusionner_v2(l1, l2):
    sentinelle = object() # un objet unique, impossible à confondre
    return [x
            for paire in zip_longest(l1, l2, fillvalue=sentinelle)
            for x in paire
            if x is not sentinelle]

print(fusionner_v2([1, 2, 3], ['a', 'b']))
````

````python
def ajouter_zero(liste):
    """Retourne une NOUVELLE liste avec un zéro à la fin.

    La liste passée en argument n'est pas modifiée.
    """
    nouvelle = liste.copy() # ou list(liste), ou liste[:]
    nouvelle.append(0)
    return nouvelle
````
